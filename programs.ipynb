```python
for i in range(1, 500):
    strong = i
    temp = i
    total = 0

    while temp > 0:
        digit = temp % 10

        fact = 1
        for j in range(1, digit + 1):
            fact = fact * j

        total = total + fact
        temp = temp // 10

    if total == strong:
        print(strong)
```

```python
n= int(input("Enter a N: "))

count = 0
num = 2

while True:
    prime = True
    
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            prime = False
            break

    if prime:
        count += 1
        if count == n:
            print(num)
            break
    num += 1        
```

```python
n=int(input('enter a number'))
for i in range(1,n):
    count=0
    sum=0
    bkp=i
    bkp1=i
    while i>0:
        count+=1
        i=i//10
    while bkp>0:
        digit=bkp%10
        sum=sum+digit**count
        bkp=bkp//10
    if sum==bkp1:
        print('armstrong number',bkp1)
```
